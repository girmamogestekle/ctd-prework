```python
employeeName = input("Enter Employee Name? ")
workedYears = int(input("Enter How many years are you working at our company? "))
if workedYears >= 5:
  print("Congrats " + employeeName + "! You have " + str(workedYears) + " years of experience, so you have a company benefits.")
else:
  print("Sorry " + employeeName + "!, you are not qualified for our benefits but you need " + str(5-workedYears)+ " years of experience to get our benefits .")
```
